```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
customer_orders = set()
```

```python
add_more = "yes"

while add_more == "yes":
    product = input("Enter the name of the product you want to order: ")

    if product in products:
        customer_orders.add(product)
        print(f"{product} added to the order.")
    else:
        print("Product not found. Try again.")

    add_more = input("Do you want to add another product? (yes/no): ")
```

```python
inventory = {'t-shirt': 9, 'mug': 5, 'hat': 7, 'book': 3, 'keychain': 2}
for item in customer_orders:
    inventory[item] -= 1

print("\nCustomer orders:", customer_orders)
print("Updated inventory:", inventory)
```
