```python
import pandas as pd
import numpy as np

insurance_df1 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv")

print(insurance_df1.info())
print()
insurance_df1.head()
```

```python
insurance_df2 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file2.csv")

print(insurance_df2.info())
print()
insurance_df2.head()
```

```python
insurance_df3 = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file3.csv")

print(insurance_df3.info())
print()
insurance_df3.head()
```

```python
# Combining all dataframes in one called "all_insurance_df"

all_insurance_df = pd.concat([insurance_df1, insurance_df2, insurance_df3], axis=0, ignore_index=True)
print(all_insurance_df.info())
print()
all_insurance_df.head()
```

```python
all_insurance_df.columns
```

```python
# Cleaning columns name and dropping dupplicated columns

all_insurance_df=all_insurance_df.rename(columns={
        "Customer": "customer",
        "ST": "state",
        "GENDER": "gender",
        "Education": "education",
        "Customer Lifetime Value": "customer_lifetime_value",
        "Income": "income",
        "Monthly Premium Auto": "monthly_premium_auto",
        "Number of Open Complaints": "number_of_open_complaints",
        "Policy Type": "policy_type",
        "Vehicle Class": "vehicle_class", 
        "Total Claim Amount": "total_claim_amount",
    })

all_insurance_df
all_insurance_df.columns
```

```python
all_insurance_df = all_insurance_df.drop(['State', 'Gender'], axis=1)
all_insurance_df.columns
```

```python
# Cleaning invalid values respecting the previous lav directions

all_insurance_df['gender'] = (
    all_insurance_df['gender']
    .str.lower()
    .replace({
        'f': 'F',
        'female': 'F',
        'femal': 'F',
        'm': 'M',
        'male': 'M'
    })
)

all_insurance_df['gender'].unique()
```

```python
all_insurance_df['state'] = all_insurance_df['state'].replace({
    'AZ': 'Arizona',
    'Cali': 'California',
    'WA': 'Washington'
})

all_insurance_df['state'].unique()
```

```python
all_insurance_df['education'] = all_insurance_df['education'].replace({
    'Bachelors': 'Bachelor'
})

all_insurance_df['education'].unique()
```

```python
all_insurance_df['customer_lifetime_value'] = (
    all_insurance_df['customer_lifetime_value']
    .str.replace('%', '', regex=False)
)

all_insurance_df['customer_lifetime_value'].unique()
```

```python
all_insurance_df['vehicle_class'] = all_insurance_df['vehicle_class'].replace({
    "Sports Car": "Luxury",
    "Luxury SUV": "Luxury",
    "Luxury Car": "Luxury"
})

all_insurance_df['vehicle_class'].unique()
```

```python
# Formatting data types
#Customer lifetime value should be numeric

all_insurance_df['customer_lifetime_value'] = (
    all_insurance_df['customer_lifetime_value']
    .str.replace('%', '', regex=False)
    .astype(float)
)

all_insurance_df['customer_lifetime_value'].sample
```

```python
# Number of open complaints has an incorrect format. 
# Look at the different values it takes with unique() and take the middle value

all_insurance_df['number_of_open_complaints'] = (
    all_insurance_df['number_of_open_complaints']
    .str.split('/')
    .str[1]
    .astype(float)
)

all_insurance_df['number_of_open_complaints'].sample
```

```python
# Dealing with Null Values

all_insurance_df[all_insurance_df.isnull().any(axis=1)]
```

```python
all_insurance_df.columns[all_insurance_df.isnull().any()]
```

```python
all_insurance_df["gender"].sample(20)
```

```python
all_insurance_df.isnull()
```

```python
# Preencher Null Values por Unknown nas colunas de texto

cols_categoricas = all_insurance_df.select_dtypes(include="object").columns
all_insurance_df[cols_categoricas] = all_insurance_df[cols_categoricas].fillna("Unknown")

cols_categoricas.isnull()
```

```python
all_insurance_df.isnull().sum()
```

```python
# "customer_lifetime_value" and "number_of_open_complaints" have a huge number of null values
# in this case, my decision is to drop both columns

all_insurance_df = all_insurance_df.drop(['customer_lifetime_value', 'number_of_open_complaints'], axis=1)
all_insurance_df.columns
```

```python
# For "income", "monthly_premium_auto" and "total_claim_amount", we will fill with the mean value

cols_numericas = all_insurance_df.select_dtypes(include="number").columns
all_insurance_df[cols_numericas] = all_insurance_df[cols_numericas].fillna(
    all_insurance_df[cols_numericas].median()
)

cols_numericas.isnull()
```

```python
#Done!!
all_insurance_df.isnull().sum()
```

# Challenge 2: Structuring Data

In this challenge, we will continue to work with customer data from an insurance company, but we will use a dataset with more columns, called marketing_customer_analysis.csv, which can be found at the following link:

https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv

This dataset contains information such as customer demographics, policy details, vehicle information, and the customer's response to the last marketing campaign. Our goal is to explore and analyze this data by performing data cleaning, formatting, and structuring.

```python
import pandas as pd
import numpy as np

mkt_insurance_df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv")

print(mkt_insurance_df.info())
print()
mkt_insurance_df.head()
```

1. You work at the marketing department and you want to know which sales channel brought the most sales in terms of total revenue. Using pivot, create a summary table showing the total revenue for each sales channel (branch, call center, web, and mail).
Round the total revenue to 2 decimal points.  Analyze the resulting table to draw insights.

```python
mkt_insurance_df.groupby("sales_channel")["monthly_premium_auto"].sum()
```

```python
pivot_revenue = (
    mkt_insurance_df.pivot_table(
        values="monthly_premium_auto",
        index="sales_channel",
        aggfunc="sum"
    )
    .round(2)
)

pivot_revenue
```

2. Create a pivot table that shows the average customer lifetime value per gender and education level. Analyze the resulting table to draw insights.

```python
pivot_average = (
    mkt_insurance_df.pivot_table(
        values="customer_lifetime_value",
        index="gender",
        columns="education",
        aggfunc="mean"
    )
    .round(2)
)

pivot_average
```

## Bonus

You work at the customer service department and you want to know which months had the highest number of complaints by policy type category. Create a summary table showing the number of complaints by policy type and month.
Show it in a long format table.

*In data analysis, a long format table is a way of structuring data in which each observation or measurement is stored in a separate row of the table. The key characteristic of a long format table is that each column represents a single variable, and each row represents a single observation of that variable.*

*More information about long and wide format tables here: https://www.statology.org/long-vs-wide-data/*

```python
complaints_summary = (
    mkt_insurance_df
        .groupby(["policy_type", "month"])["number_of_open_complaints"]
        .count()
        .reset_index()
)

complaints_summary
```
