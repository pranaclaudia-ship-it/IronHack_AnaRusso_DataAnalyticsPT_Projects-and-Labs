```python
import pandas as pd
import numpy as np
import re
```

```python
marketing_df = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv")
marketing_df.head()
```

```python
marketing_df.columns
```

```python
#Step1 - Create a new DataFrame that only includes customers who:
# have a low total_claim_amount (e.g., below $1,000),
# have a response "Yes" to the last marketing campaign

filtered_df = marketing_df[
    (marketing_df["Response"] == "Yes") &
    (marketing_df["Total Claim Amount"] <= 1000)
]

filtered_df.head()
```

```python
#Step 2 - Using the original Dataframe, analyze:
# the average monthly_premium and/or customer lifetime value by policy_type and gender for customers who responded "Yes", and compare
# these insights to total_claim_amount patterns, and discuss which segments appear most profitable or low-risk for the company

yes_df = marketing_df[marketing_df["Response"] == "Yes"]
```

```python
# average monthly_premium
premium_avg = yes_df.groupby(["Policy Type", "Gender"])["Monthly Premium Auto"].mean()
premium_avg
```

```python
#total_claim_amount patterns
claims_avg = yes_df.groupby(["Policy Type", "Gender"])["Total Claim Amount"].mean()
claims_avg
```

```python
agg_df = yes_df.groupby(["Policy Type", "Gender"]).agg({
    "Monthly Premium Auto": "mean",
    "Customer Lifetime Value": "mean",
    "Total Claim Amount": "mean"
})

agg_df
```

```python
#Step 3: Analyze the total number of customers who have policies in each state, 
#and then filter the results to only include states where there are more than 500 customers.

customers_by_state = marketing_df.groupby("State")["Customer"].count()
customers_by_state
```

```python
big_states = customers_by_state[customers_by_state > 500]
big_states
```

```python
#Step 4: Find the maximum, minimum, and median customer lifetime value by education level and gender. Write your conclusions
clv_stats = marketing_df.groupby(["Education", "Gender"])["Customer Lifetime Value"].agg(["max", "min", "median"])
clv_stats
```
